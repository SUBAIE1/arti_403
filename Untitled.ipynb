```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from numpy.fft import fft2, ifft2, ifftshift

img = cv2.imread('image.png', cv2.IMREAD_GRAYSCALE)
img = cv2.resize(img, (256, 256))
img = img.astype(np.float64)

# Sobel filters
sobel_x = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
])

sobel_y = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
])

# Center and embed the kernel, then compute its FFT
def center_embed_kernel(kernel, shape):
    padded = np.zeros(shape)
    kh, kw = kernel.shape
    ph, pw = shape

    cy, cx = ph // 2, pw // 2

    padded[cy - kh//2:cy - kh//2 + kh,
           cx - kw//2:cx - kw//2 + kw] = kernel

    return fft2(ifftshift(padded))

# Compute FFT of image
F = fft2(img)

# FFT of Sobel filters
H_x = center_embed_kernel(sobel_x, img.shape)
H_y = center_embed_kernel(sobel_y, img.shape)

# Multiply in the frequency domain
G_x = F * H_x
G_y = F * H_y

# Convert back to spatial domain
result_x = np.real(ifft2(G_x))
result_y = np.real(ifft2(G_y))

# Gradient magnitude
magnitude = np.sqrt(result_x**2 + result_y**2)

# Display results
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Cameraman Image')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(np.abs(result_x), cmap='gray')
plt.title('Sobel X - Frequency Domain')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(np.abs(result_y), cmap='gray')
plt.title('Sobel Y - Frequency Domain')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(magnitude, cmap='gray')
plt.title('Sobel Gradient Magnitude')
plt.axis('off')

plt.tight_layout()
plt.show()
```
